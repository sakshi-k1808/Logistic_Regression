```python
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler()
dl=['Pregnancies','Outcome']
data1=sc.fit_transform(data.drop(dl,axis=1))
```

```python
data1
```

```python
con_data=data[['Pregnancies','Outcome']]
con_data
```

```python
data.columns
```

```python
print(type(data1))
```

```python
data2=pd.DataFrame(data1,columns=['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age'])

data2
```

```python
final_df=pd.concat([data2,con_data],axis=1)
final_df
```

```python
final_df.describe()
```

```python
sns.heatmap(data2.corr(),annot=True)
plt.show()
```

```python
final_df.head()
```

```python
X=final_df.iloc[:,:-1]
y=final_df.Outcome
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=45)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression
clf=LogisticRegression()
clf.fit(X_train,y_train)
```

```python
y_pred=clf.predict(X_test)
y_pred
```

```python
y_test
```

```python
y_pred_prob=clf.predict_proba(X_test)
```
